# Fasta database overview

Mixture FASTA: /public/conode53/sam/SCI/fasta/human_ecoli_yeast_contaminant_revi_2025_10_17.fasta

Human FASTA: /public/conode53/sam/SCI/fasta/human_contaminant_revi_2025_10_17.fasta

Contaminant sequences have Cont_ as prefix.

```python
from pathlib import Path
import re

FASTA_DIR = Path("/public/conode53/sam/SCI/fasta")

MIXTURE_FASTA = FASTA_DIR / "human_ecoli_yeast_contaminant_revi_2025_10_17.fasta"
HUMAN_FASTA   = FASTA_DIR / "human_contaminant_revi_2025_10_17.fasta"

def parse_fasta_headers(path: Path) -> list[str]:
    headers = []
    with open(path) as f:
        for line in f:
            if line.startswith(">"):
                headers.append(line.strip())
    return headers

mixture_headers = parse_fasta_headers(MIXTURE_FASTA)
human_headers   = parse_fasta_headers(HUMAN_FASTA)

print(f"Mixture FASTA: {len(mixture_headers):,} sequences")
print(f"Human FASTA:   {len(human_headers):,} sequences")
```

```python
def classify_headers(headers: list[str]) -> dict:
    ox_re = re.compile(r"OX=(\d+)")
    stats = {"contaminant": 0, "human": 0, "ecoli": 0, "yeast": 0, "other": 0}
    for h in headers:
        if "Cont_" in h:
            stats["contaminant"] += 1
            continue
        m = ox_re.search(h)
        ox = m.group(1) if m else None
        if ox == "9606":
            stats["human"] += 1
        elif ox == "83333":
            stats["ecoli"] += 1
        elif ox == "559292":
            stats["yeast"] += 1
        else:
            stats["other"] += 1
    return stats

mixture_stats = classify_headers(mixture_headers)
human_stats   = classify_headers(human_headers)

print("=== Mixture FASTA ===")
for k, v in mixture_stats.items():
    print(f"  {k:>12s}: {v:,}")

print("\n=== Human FASTA ===")
for k, v in human_stats.items():
    print(f"  {k:>12s}: {v:,}")
```
